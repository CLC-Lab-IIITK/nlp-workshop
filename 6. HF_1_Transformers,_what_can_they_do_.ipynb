## Pipeline
**Pipelines** are objects that abstract most of the complex code from the library, offering a simple API dedicated to several tasks

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier("I've been waiting for a HuggingFace course my whole life.")
```

```python
sentiment = classifier ("The script of the movie is so impressive though the casting is poor.")
sentiment[0]
```

```python
classifier(
    ["I've been waiting for a HuggingFace course my whole life.", "I hate this so much!"]
)
```

```python
classifier(["The movie was watchable though the story is lagging","The movie not was watchable"])
```

```python
from transformers import pipeline

question_answerer = pipeline("question-answering")
question_answerer(
    question="Where do I work?",
    context="My name is Sylvain and I work at Hugging Face in Brooklyn",
)
```

```python
## this is wrong!!!
question_answerer(
    question="Who is my employer?",
    context="My name is Sylvain and I work at Hugging Face in Brooklyn",
)
```

```python
context = "Deep learning is a method in artificial intelligence (AI) that teaches computers to process data in a way that is inspired by the human brain. Deep learning models can recognize complex patterns in pictures, text, sounds, and other data to produce accurate insights and predictions."
ans= question_answerer(
    question="What are the applications of Deep Learning?",
    context=context,
)
ans['answer']
```
