```python
corpus = [
    "I love natural language processing",
    "I love machine learning",
    "I love data science",
    "I like natural language processing",
    "I like machine learning",
    "I love natural products"
    "I study data science"
]
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(corpus[0].lower())
print(tokens)
```

```python
from nltk.util import ngrams
from collections import Counter

text = " ".join(corpus).lower()

tokens = word_tokenize(text)

bigrams = list(ngrams(tokens, 2))

bigram_counts = Counter(bigrams)

print(bigram_counts)
```

```python
from collections import Counter

unigrams = Counter(tokens)
bigrams = Counter(ngrams(tokens, 2))

def bigram_probability(w1, w2):
    return bigrams[(w1, w2)] / unigrams[w1]

print("P(machine|love)= ",bigram_probability("love", "machine"))
print("P(natural|love)= ",bigram_probability("love", "natural"))
print("P(data|love)= ",bigram_probability("love", "data"))
```

```python
def predict_next_word(word):
    candidates = []

    for (w1, w2), count in bigrams.items():
        if w1 == word:
            candidates.append((w2, count))

    candidates.sort(key=lambda x: x[1], reverse=True)

    return candidates
```

```python
print(predict_next_word("love")[0][0])
```

```python
next = "i"
for i in range(5):
  print(next)
  next = predict_next_word(next)[0][0]
```
